# Transcription Factors Enrichment Analysis

ULM incorporates weights of TF-gene (i.e. source-target) interactions.

The scores from decoupler.mt.ulm() should be interpreted such that larger magnitudes indicate greater significance, while the sign reflects whether the features in the set are overrepresented (positive) or underrepresented (negative) compared to the background.

Recall: the enrichment analysis using Univariate Linear Model (ulm) fits a linear predictor to the genes expression profile of each sample using as predictors the interaction weights of the genes targets of TFs. Then it applies a regression t-test to check whether the fitted line is good. In other words, if the TF whose targets approximate the gene expression profile is active. The t value is the Enrichment Score (ES). Positive ES means enrichment of TF, depletion otherwise. 

https://decoupler.readthedocs.io/en/latest/notebooks/scell/rna_sc.html

### Results processing: select TF significantly enriched in more than 1/4 of cells

```python
########################################################
### Select some TFS ###
## Select only significant enrichment scores ##
########################################################


#### Parameters ####

# Significance threshold
alpha = 0.05
# Ratio of cells from an anatomical cluster in which the TF must be significant to kept
ratio = 1/4 #e.g. if 1/2, it means we keep only TFs that are significant in at least half of cells of a given anatomical cluster

######################



padj_df_dorothea = adata_muscle_hvg.obsm["padj_ulm_dorothea"]
padj_df_collectri = adata_muscle_hvg.obsm["padj_ulm_collectri"]

# Convert padj_df to long format
padj_df_dorothea["id"]=padj_df_dorothea.index
padj_df_dorothea_long=pd.melt(padj_df_dorothea, id_vars=["id"], var_name="TF", value_name="padj")

padj_df_collectri["id"]=padj_df_collectri.index
padj_df_collectri_long=pd.melt(padj_df_collectri, id_vars=["id"], var_name="TF", value_name="padj")

# Add a column anatomical_cluster based on id column and its corresponding "anatomical_cluster" value in adata_muscle
cluster_map = adata_muscle_hvg.obs.set_index("id")["anatomical_cluster"]
padj_df_dorothea_long["anatomical_cluster"] = padj_df_dorothea_long["id"].map(cluster_map)
padj_df_collectri_long["anatomical_cluster"] = padj_df_collectri_long["id"].map(cluster_map)

# Flag significant values
padj_df_dorothea_long["significant"] = padj_df_dorothea_long["padj"] <= alpha
padj_df_collectri_long["significant"] = padj_df_collectri_long["padj"] <= alpha

# Compute for each (anatomical cluster, TF) how many are significant
significance_counts_dorothea = (
    padj_df_dorothea_long[padj_df_dorothea_long["significant"]]
    .groupby(["anatomical_cluster", "TF"])
    .size()
    .rename("significant_count")
    .reset_index()
)
significance_counts_collectri = (
    padj_df_collectri_long[padj_df_collectri_long["significant"]]
    .groupby(["anatomical_cluster", "TF"])
    .size()
    .rename("significant_count")
    .reset_index()
)

# Compute anatomical cluster sizes (number of unique ids in each cluster)
cluster_sizes_dorothea = (
    padj_df_dorothea_long[["id", "anatomical_cluster"]]
    .drop_duplicates()
    .groupby("anatomical_cluster")
    .size()
    .rename("cluster_size")
    .reset_index()
)
cluster_sizes_collectri = (
    padj_df_collectri_long[["id", "anatomical_cluster"]]
    .drop_duplicates()
    .groupby("anatomical_cluster")
    .size()
    .rename("cluster_size")
    .reset_index()
)

# Add the cluster size info to significance counts
significance_counts_dorothea = significance_counts_dorothea.merge(cluster_sizes_dorothea, on="anatomical_cluster")
significance_counts_collectri = significance_counts_collectri.merge(cluster_sizes_collectri, on="anatomical_cluster")

# Keep only TFs where significant_count >= cluster_size * ratio i.e. TF significantly enriched/depleted in more than ratio of cells of the anatomical cluster
selected_TF_df_dorothea = significance_counts_dorothea[
    significance_counts_dorothea["significant_count"] >= significance_counts_dorothea["cluster_size"] * ratio
]
selected_TF_df_collectri = significance_counts_collectri[
    significance_counts_collectri["significant_count"] >= significance_counts_collectri["cluster_size"] * ratio
]

selected_TF_list_dorothea = selected_TF_df_dorothea["TF"].unique()
selected_TF_list_collectri = selected_TF_df_collectri["TF"].unique()
```

# Differential Expression Analysis 

```python
#### Aggregate several TF databases, to label in gold TF in DEA dotplots ####
collectri = dc.op.collectri(organism="human")
dorothea = dc.op.dorothea(levels=['A', 'B', 'C', 'D'], organism="human")

db_1 = collectri['source'].to_list()
db_2 = dorothea['source'].to_list()
db = db_1 + db_2
TF_genes = list(set(db))
```

DEA cannot be done with DESeq2 because data is too sparse, throwing error when DESeq2 tries to estimate size factors (because it relies on geometric means, and thus on log but log(0) is not defined).

Rather apply scanpy method to DEA, using t-test and/or wilcoxon.

## Cluster1 versus. rest (union of every cells not in cluster1)

```python
sc.tl.rank_genes_groups(adata_muscle_hvg, groupby="anatomical_cluster", method="t-test", use_raw=False, layer="log_norm") #it expects log normalized data, so I set use_raw to False 
adata_muscle_hvg
```

### Dotplots

```python
anatomical_clusters = ["EOM", "arches", "tongue", "above_tongue", "masseter"]

for cluster in anatomical_clusters:
    ## Filter for only significant DEA genes ##
    df_muscle_only = sc.get.rank_genes_groups_df(adata_muscle_hvg, group=cluster)
    df_muscle_only = df_muscle_only[df_muscle_only['pvals_adj'] < 0.05]

    # Order to have higher logFC at the top and only positive logFC i.e. upregulated genes
    df_muscle_only = df_muscle_only.sort_values(by='logfoldchanges', ascending=False)
    df_muscle_only = df_muscle_only[df_muscle_only['logfoldchanges']>0]

    # Dotplot of upregulated genes
    dotplot = sc.pl.dotplot(adata_muscle_hvg, var_names=df_muscle_only['names'], groupby='anatomical_cluster', show=False, swap_axes=True) 
    ax = dotplot["mainplot_ax"]
    ax.set_title(f"Upregulated genes in {cluster}", fontsize=12, pad=10)


    for label in ax.get_yticklabels():
        label.set_style("italic")
        gene = label.get_text()
        if gene in TF_genes:
            label.set_color("#A97F03")
            label.set_fontweight("bold")
        else:
            label.set_color("black")
            label.set_fontweight("normal")
    plt.show()
```

## Pairwise Comparisons

```python
#### Perform Pairwise DEA ####
anatomical_clusters = ["EOM", "arches", "tongue", "above_tongue", "masseter"]
results = {}

for i in range(len(anatomical_clusters)):
    for j in range(i+1, len(anatomical_clusters)):
        group = anatomical_clusters[i]
        reference = anatomical_clusters[j]
        
        sc.tl.rank_genes_groups(
            adata_muscle_hvg,
            groupby='anatomical_cluster',
            groups=[group],
            reference=reference,
            method='t-test',  # "wilcoxon"
            n_genes=adata_muscle_hvg.shape[1],
            layer="log_norm",
            use_raw=False, 
            copy=False,
            key_added=f"rank_{group}_vs_{reference}"  # store results under custom keys
        )
        
        # Collect results
        results[(group, reference)] = adata_muscle_hvg.uns[f"rank_{group}_vs_{reference}"]
```

Note: I get this warning but I made sure I am using the log normalized data as input, by setting use_raw=False and layer=counts (where I previously stored adata.X to layer "counts" and this was indeed log normalized in nuclei_segmentation.ipynb).

```python
#### Plot Results ####

# And save the genes, for example to study them at the megabulk level
path = "./single_cell_DEA_results"

for (group, reference), res in results.items():
    key = f"rank_{group}_vs_{reference}"
    df_pairwise = sc.get.rank_genes_groups_df(adata_muscle_hvg, group=group, key=key)

    ## Genes Upregulated in group ##
    df_pairwise_up_group1 = df_pairwise[df_pairwise['pvals_adj'] < 0.05]
    df_pairwise_up_group1 = df_pairwise_up_group1[df_pairwise_up_group1['logfoldchanges'] > 0]
    df_pairwise_up_group1 = df_pairwise_up_group1.sort_values(by='logfoldchanges', ascending=False).head(70) # top 70 upregulated genes
    # saving as tsv file
    df_pairwise_up_group1.to_csv(os.path.join(path,f"up_in_{group}_vs_{reference}.tsv"), sep="\t")
    
    ## Genes Upregulated in reference ##
    df_pairwise_up_group2 = df_pairwise[df_pairwise['pvals_adj'] < 0.05]
    df_pairwise_up_group2 = df_pairwise_up_group2[df_pairwise_up_group2['logfoldchanges'] < 0]  
    df_pairwise_up_group2 = df_pairwise_up_group2.sort_values(by='logfoldchanges')  # more negative on top
    df_pairwise_up_group2 = df_pairwise_up_group2.head(70)  # top 70 upregulated genes
    df_pairwise_up_group2.to_csv(os.path.join(path,f"up_in_{reference}_vs_{group}.tsv"), sep="\t")
    
    if df_pairwise_up_group1.empty:
        print(f"No significant upregulated genes for {group} vs {reference}")
        continue

    if df_pairwise_up_group2.empty:
        print(f"No significant upregulated genes for {reference} vs {group}")
        continue
    
    heatmap_1=sc.pl.heatmap(
        adata=adata_muscle_hvg,
        use_raw=False,
        groupby="anatomical_cluster",
        var_names=df_pairwise_up_group1['names'],
        cmap="gnuplot_r",
        figsize=(8,10),
        show=False,
        show_gene_labels=True
    )
    
    ax_1 = heatmap_1["heatmap_ax"]
    ax_1.set_title(f"Upregulated genes in {group} vs {reference}", fontsize=12, pad=10)
    
    for label in ax_1.get_yticklabels():
        label.set_style("italic")
        gene = label.get_text()
        if gene in TF_genes:
            label.set_color("#A97F03")
            label.set_fontweight("bold")
        else:
            label.set_color("black")
            label.set_fontweight("normal")

    heatmap_2=sc.pl.heatmap(
        adata=adata_muscle_hvg,
        use_raw=False,
        groupby="anatomical_cluster",
        var_names=df_pairwise_up_group2['names'],
        cmap="gnuplot_r",
        figsize=(8,10),
        show=False,
        show_gene_labels=True
    )
    
    ax_2 = heatmap_2["heatmap_ax"]
    ax_2.set_title(f"Upregulated genes in {reference} vs {group}", fontsize=12, pad=10)
    
    for label in ax_2.get_yticklabels():
        label.set_style("italic")
        gene = label.get_text()
        if gene in TF_genes:
            label.set_color("#A97F03")
            label.set_fontweight("bold")
        else:
            label.set_color("black")
            label.set_fontweight("normal")
    
    plt.show()
```
